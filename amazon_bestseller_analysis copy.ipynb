```python
import pandas as pd
import plotly.graph_objects as go
import numpy as np
```

```python
data = pd.read_csv('bestsellers with categories.csv')  # adjust path if needed
data.head()
```

# Genre Distribution

We visualize the overall distribution of books by genre using a donut pie chart. This helps us understand which genre dominates the bestseller lists.

- **Fiction vs. Non-Fiction**: Percentage and count of unique books in each category.



```python
data = data.drop_duplicates('Name')

True_labels = data['Genre'].drop_duplicates()

g_count = data['Genre'].value_counts()

fig = go.Figure(
    go.Pie(
        labels=g_count.index,
        values=g_count.values,
        hole=0.45,  # Donut chart
        textinfo='percent+label',  # show both label and percent
        hoverinfo='label+percent+value',  # hover shows counts too
        marker=dict(colors=['#1f77b4', '#ff7f0e'], line=dict(color='white', width=2)),
        pull=[0.02, 0.02],  # slightly pull slices apart for effect
    )
)

# Layout styling
fig.update_layout(
    title_text='Distribution of Genre for Unique Books (2009-2019)',
    title_font=dict(size=22, family='monospace', weight= 'bold'),
    title_x=0.5,
    annotations=[dict(text='Genre', x=0.5, y=0.5, font_size=20, showarrow=False)],
    showlegend=True,
    legend=dict(title='Genre', font=dict(size=14)),
    paper_bgcolor='RGB(240, 234, 214)'
)

fig.show()
```

# Yearly Genre Distribution

We analyze how the distribution of genres changes each year (2009–2019) using an interactive dropdown pie chart.

- Allows comparison of genre trends over time.
- Highlights years dominated by a specific genre.


```python
years = np.sort(data['Year'].unique())
```

```python
figure = go.Figure()

for year in years:
    your = data[data['Year'] == year]
    counts = your['Genre'].value_counts()

    figure.add_trace(
        go.Pie(
            labels=counts.index,
            values= counts.values,
            name=str(year),
            hole=0.45,
            textinfo='percent+label',
            hoverinfo='label+percent+value',
            visible=(year == years[0]),  # show only first year initially
            marker=dict(colors=['#1f77b4', '#ff7f0e'])
        )
    )

figure.update_layout(
    updatemenus=[dict(
        buttons=[dict(label=str(year),
                      method='update',
                      args=[{'visible': [y==year for y in years]},
                            {'title': f'Genre Distribution in {year}'}])
                 for year in years],
        direction='down',
        showactive=True
    )],
    title_text=f'Genre Distribution in {years[0]}',
    title_font= dict(size=22, family='monospace', weight= 'bold'),
    title_x= 0.5,
    paper_bgcolor='RGB(240, 234, 214)'
)

figure.show()
```

# Top Authors by Genre

We identify the most prolific authors in each genre:

- **Fiction Authors**: Top 10 authors with most books.
- **Non-Fiction Authors**: Top 10 authors with most books.

Bar charts visualize the counts for easier comparison.


```python
top_nf_auth = data[data['Genre']=='Non Fiction'].groupby('Author')['Name'].count().sort_values(ascending=True).tail(10)
top_f_auth  = data[data['Genre']=='Fiction'].groupby('Author')['Name'].count().sort_values(ascending=True).tail(10)

print(top_nf_auth.index)
print(top_nf_auth.values)
print(top_f_auth.index)
print(top_f_auth.values)
```

```python

non_fic = go.Figure(
    data=go.Bar(
        x=top_nf_auth.index,  # Bar charts use x for categories
        y=top_nf_auth.values,
        text=top_nf_auth.values,
        textposition='auto',  # shows values on top of bars
        hoverinfo='x+y'
    )
)

non_fic.update_layout(
    title_text='Top Non Fiction Authors (2009-2019)',
    title_font=dict(size=22, family='monospace', weight='bold'),
    title_x=0.5
)

non_fic.show()
```

```python
fic = go.Figure(
    data=go.Bar(
        x=top_f_auth.index,  # Bar charts use x for categories
        y=top_f_auth.values,
        text=top_f_auth.values,
        textposition='auto',  # shows values on top of bars
        hoverinfo='x+y'
    )
)

fic.update_layout(
    title_text='Top Fiction Authors (2009-2019)',
    title_font=dict(size=22, family='monospace', weight='bold'),
    title_x=0.5
)

fic.show()
```

# Highest-Rated Books

We extract the top 10 books with the highest user ratings in each genre:

- **Fiction**: Books with outstanding ratings among fiction bestsellers.
- **Non-Fiction**: Books with highest ratings in non-fiction.

This highlights critically acclaimed books that may not be the most prolific.


```python
# Top 10 highest-rated Non-Fiction
top_book_nf = pd.DataFrame(
    data[data['Genre'] == 'Non Fiction']
    .sort_values(by=['User Rating'], ascending=False)
    [['Name','User Rating']]
    .head(10)
)

# Top 10 highest-rated Fiction
top_book_f = pd.DataFrame(
    data[data['Genre'] == 'Fiction']
    .sort_values(by=['User Rating'], ascending=False)
    [['Name','User Rating']]
    .head(10)
)
```

# Most Frequent Authors

We analyze the authors who appear most often across the bestseller list (2009–2019):

- Top 15 authors overall.
- Helps identify consistent performers in the book market.


```python
# Count author appearances
appearances = data['Author'].value_counts().head(15)

# Plot as bar chart
appear = go.Figure(
    data=go.Bar(
        x=appearances.values,
        y=appearances.index,
        orientation='h',
        marker=dict(color='teal')
    )
)

appear.update_layout(
    title_text='Top 15 Authors by Number of Appearances',
    title_font=dict(size=22, family='monospace', weight='bold'),
    xaxis_title='Number of Books',
    yaxis_title='Author',
    yaxis=dict(autorange='reversed'),
    paper_bgcolor='RGB(240, 234, 214)',
    plot_bgcolor='RGB(196, 188, 180)'  # keep highest at top
)

appear.show()
```
